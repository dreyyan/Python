```python
import numpy as np
import matplotlib.pyplot as plt

# Define the range of x values (days)
t = np.linspace(0, 100, 1000)

# Calculate y values using the linear equation y = 0.1x + 1
x = 0.1 * t + 1

# Create the plot
fig, ax = plt.subplots()

plt.title("Energy (kJ)/day")
plt.xlabel("Days since April 1st")
plt.ylabel("Energy (kJ/day)")

ax.set_xlim(0, 100)
ax.set_ylim(0, 100)

ax.plot(t, x, c='green', label='Power Generation')

# Add vertical line at day 30
plt.axvline(x=30, color='purple', linestyle='--', label='May 1st')

plt.legend()
plt.grid(True)
plt.show()
```
