```python
import numpy as np
import pandas as pd
import re

df = pd.read_csv('employee_records[0].csv')

# Name
df['Name']=df['Name'].fillna(np.nan)
df['Name']=df['Name'].str.capitalize()

# Age
df['Age']=df['Age'].fillna(df['Age'].median())
df['Age']=df['Age'].round(0).astype(int)

# Gender
df['Gender']=df['Gender'].str.lower()
df['Gender']=df['Gender'].map({'female': 'F', 'male': 'M'})

# Email
email_pattern = r"^(?!.*\.\.)[\w\.-]+@[\w-]+(\.[\w-]+)+$"
df['Email']=df['Email'].apply(lambda x: x if re.match(email_pattern, str(x)) else np.nan)

# Salary
df['Salary']=df['Salary'].fillna(df['Salary'].median())
df['Salary']=df['Salary'].round(0).astype(int)

# JoinDate
df['JoinDate']=df['JoinDate'].str.replace('/', '-')
df['JoinDate']=pd.to_datetime(df['JoinDate'], errors='coerce')

# Department
df['Department']=df['Department'].str.capitalize()
df=pd.get_dummies(df, columns=['Department'], drop_first=True)

df
```
