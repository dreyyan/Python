```python
import pandas as pd
import joblib
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.svm import SVR

# load & prepare data
data = pd.read_csv('../../CSV/regression_dataset.csv')
X = data.drop(columns='target')
y = data['target']

print("[ DATA ]")
print(data)

# split dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# train model
model = SVR()
model.fit(X_train, y_train)

# test model
predictions = model.predict(X_test)

# evaluate model
mae = mean_absolute_error(y_test, predictions)
mse = mean_squared_error(y_test, predictions)
rmse = mse ** 0.5
r2 = r2_score(y_test, predictions)

# save model
joblib.dump(model, 'models/SVM_regressor.joblib')

# 1. R² Score - Bar Plot
print("[ R² SCORE ]")
plt.figure(figsize=(4, 4))
plt.bar(["R² Score"], [r2], color='skyblue')
plt.ylim(0, 1)
plt.title("Model R² Score")
plt.ylabel("Score")
plt.show()

# 2. Actual vs Predicted - Scatter Plot
plt.figure(figsize=(8, 4))
sns.scatterplot(x=y_test, y=predictions, color='purple')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color='red', linestyle='--')
plt.title("Actual vs Predicted")
plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.tight_layout()
plt.show()

# 3. Print Error Metrics
print(f"Mean Absolute Error: {mae:.2f}")
print(f"Mean Squared Error: {mse:.2f}")
print(f"Root Mean Squared Error: {rmse:.2f}")
print(f"R² Score: {r2:.2f}")

# 4. Cross-Validation Scores - R²
scores = cross_val_score(model, X, y, cv=5, scoring='r2')
plt.figure(figsize=(8, 4))
plt.bar(range(1, 6), scores, color='lightgreen')
plt.ylim(0, 1)
plt.title("Cross-Validation R² Score per Fold")
plt.xlabel("Fold")
plt.ylabel("R² Score")
plt.xticks(range(1, 6))
plt.tight_layout()
plt.show()
```
