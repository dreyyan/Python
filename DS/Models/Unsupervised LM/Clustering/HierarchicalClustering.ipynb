```python
from sklearn.datasets import make_blobs
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as sch
import numpy as np
# for simple, spherical blobs

# Step 1: Generate blob-shaped data (ideal for hierarchical clustering)
X, _ = make_blobs(n_samples=300, centers=3, cluster_std=1.0, random_state=42)

# Step 2: Plot dendrogram to visualize cluster merging
plt.figure(figsize=(10, 5))
dendrogram = sch.dendrogram(sch.linkage(X, method='ward'))
plt.title("Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Euclidean Distance")
plt.grid(True)
plt.show()

# Step 3: Apply Hierarchical Clustering
cluster = AgglomerativeClustering(n_clusters=3, linkage='ward')
labels = cluster.fit_predict(X)

# Step 4: Visualize result
plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='rainbow', s=30)
plt.title("Agglomerative Clustering on Blob Data")
plt.xlabel("X Position")
plt.ylabel("Y Position")
plt.grid(True)
plt.show()
```
