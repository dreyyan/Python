```python
''' ACTIVATION FUNCTIONS
======================================================
1. Sigmoid (Logistic)
  R - (0, 1)
[/] - old networks, binary classification
[x] - vanishing gradients
------------------------------------------------------
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)

======================================================
2. Hyperbolic Tangent (tanh)
  R - (-1, 1)
[/] - centered at 0, better than sigmoid
[x] - still suffers from vanishing gradients
------------------------------------------------------
def tanh(x):
    return np.tanh(x)

def tanh_derivative(x):
    return 1 - np.tanh(x) ** 2

======================================================
3. ReLU (Rectified Linear Unit)
  R - [0, ∞)
[/] - most popular in hidden layers
[x] - "Dying ReLU" problem (neurons stuck at 0)
------------------------------------------------------
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)
    
======================================================
4. Leaky ReLU
  R - (-∞, ∞)
[/] - fixes dying ReLU by letting negative values leak
[x] - slope α must be chosen (usually 0.01)
------------------------------------------------------
def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

def leaky_relu_derivative(x, alpha=0.01):
    return np.where(x > 0, 1, alpha)

======================================================
5. Parametric ReLU (PReLU)
  R - (-∞, ∞)
[/] - learns α (slope) during training
[x] - adds parameters, risk of overfitting
------------------------------------------------------
def prelu(x, alpha):
    return np.where(x > 0, x, alpha * x)

def prelu_derivative(x, alpha):
    return np.where(x > 0, 1, alpha)

======================================================
6. Exponential Linear Unit (ELU)
  R - (-α, ∞)
[/] - smooth, reduces bias shift
[x] - more expensive than ReLU
------------------------------------------------------
def elu(x, alpha=1.0):
    return np.where(x > 0, x, alpha * (np.exp(x) - 1))

def elu_derivative(x, alpha=1.0):
    return np.where(x > 0, 1, elu(x, alpha) + alpha)

======================================================
7. Softmax
  R - (0, 1), sums to 1
[/] - final layer for multi-class classification
[x] - derivative is full Jacobian, can be expensive
------------------------------------------------------
def softmax(x):
    exp_x = np.exp(x - np.max(x))  # stability trick
    return exp_x / np.sum(exp_x, axis=0)

# Derivative (Jacobian matrix):
# dσ_i/dx_j = σ_i * (δ_ij - σ_j)

======================================================
8. Swish
  R - (-∞, ∞)
[/] - smooth ReLU version, often outperforms it
[x] - more expensive than ReLU
------------------------------------------------------
def swish(x):
    return x * sigmoid(x)

def swish_derivative(x):
    s = sigmoid(x)
    return s + x * s * (1 - s)

======================================================
9. Gaussian Error Linear Unit (GELU)
  R - (-∞, ∞)
[/] - smooth like Swish, performs well in NLP (Transformers)
[x] - slightly more expensive to compute
------------------------------------------------------
def gelu(x):
    return 0.5 * x * (1 + np.tanh(np.sqrt(2/np.pi) * (x + 0.044715 * x**3)))

======================================================
10. Maxout
  R - (-∞, ∞)
[/] - learns piecewise linear activation
[x] - doubles parameters, heavier model
------------------------------------------------------
def maxout(x, W, b):
    # W, b = list of weight & bias pairs
    # Example: 2 linear pieces
    z1 = np.dot(x, W[0]) + b[0]
    z2 = np.dot(x, W[1]) + b[1]
    return np.maximum(z1, z2)

# Derivative depends on which branch is active
```
