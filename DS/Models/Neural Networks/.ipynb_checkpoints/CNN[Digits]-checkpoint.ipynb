```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# load dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# normalize images to [0, 1] for faster training and accuracy
x_train, x_test = x_train / 255.0, x_test / 255.0

# reshape to add a channel distribution (batch_size, height, width, channels)
# -1 - figure batch size automatically
# 28, 28 - image dimensions
# 1 - grayscale channel (3 for RGB)

x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

# print("Training set shape: ", x_train.shape)

# build CNN model
model = models.Sequential([
  # detects edges & patterns
  layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
  # reduces size & noise
  layers.MaxPooling2D((2, 2)),

  # learns deeper features
  layers.Conv2D(64, (3, 3), activation='relu'),
  # further downsamples
  layers.MaxPooling2D((2, 2)),

  # prepares for classification
  layers.Flatten(),
  # learns relationshiops
  layers.Dense(64, activation='relu'),
  # outputs probabilities per digit
  layers.Dense(10, activation='softmax') # 10 classes (0–9)
])

# compile model
model.compile(optimizer='adam', # advanced optimization algorithm
              loss='sparse_categorical_crossentropy', # instructions to measure error
              metrics=['accuracy']) # metric to track

# train model
history = model.fit(x_train, y_train, epochs=5, # train datset 5 times
                    validation_data=(x_test, y_test))

# evaluate model
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f"Test accuracy: {test_acc:.2f}")

# evaluate model by making predictions
predictions = model.predict(x_test)

# Show a few predictions
for i in range(5):
    plt.imshow(x_test[i].reshape(28, 28), cmap='gray')
    plt.title(f"Predicted: {np.argmax(predictions[i])}, Actual: {y_test[i]}")
    plt.show()
```
